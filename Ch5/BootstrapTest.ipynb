```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
```

```python
arr = np.array([6.20, 4.34, 8.14, 6.24, 3.72, 3.54, 4.35, 2.67, 7.16, 6.00])
```

```python
print("Wartosc oczekiwana: ", np.average(arr))
arr_mean = arr.mean()
n = len(arr)
n_iter = 1000

arr_means = []

for k in range(n_iter):
    arr1 = np.random.choice(arr, n, replace=True)
    arr_means.append(arr1.mean())
    
arr_means = np.array(arr_means)
arr_means.sort()

print(arr_means)
```

```python
lower, upper = np.percentile(arr_means, [5, 95])

print("95% przedzial ufnosci", lower,":", upper)
print("Czy 0? : ",lower <= 0 <= upper)
```

```python
counter = np.sum(arr_means > arr_mean)
n = len(arr_means)

ratio = counter / n
print(ratio)
```
