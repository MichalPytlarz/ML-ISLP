### Regresja liniowa opiera się na kilku założeniach. Jeśli są one naruszone, wyniki mogą być niewiarygodne.


Liniowość

Założenie: związek między zmiennymi niezależnymi a zależną jest liniowy.

Problem: jeśli zależność jest nieliniowa, model nie dopasuje się dobrze → przewidywania są błędne.

Niezależność obserwacji

Założenie: obserwacje są niezależne od siebie.

Problem: w danych czasowych lub panelowych często występuje autokorelacja (np. szereg czasowy), co narusza to założenie.

Homoskedastyczność (stała wariancja błędów)

Założenie: rozproszenie błędów jest stałe dla wszystkich wartości zmiennych niezależnych.

Problem: heteroskedastyczność (zmienna wariancja błędów) może powodować niespójne estymatory wariancji i błędne wnioski statystyczne.

Normalność reszt

Założenie: reszty (błędy) są normalnie rozłożone.

Problem: duże odchylenia od normalności mogą wpływać na testy istotności współczynników i przedziały ufności.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t, norm, chi2, f

# --- Parametry ---
n = 10              # liczba obserwacji
df = n - 2          # stopnie swobody (n - 2)
x = np.linspace(-5, 5, 500)  # zakres wartości na osi X

# --- Gęstości rozkładów ---
t_pdf = t.pdf(x, df)          # rozkład t-Studenta
norm_pdf = norm.pdf(x, 0, 1)  # rozkład normalny (średnia 0, sd=1)
chi2_pdf = chi2.pdf(x, df)
f_pdf = f.pdf(x, df, n)  # W skrócie: F mierzy względną wariancję między dwoma zestawami danych

# --- Rysowanie ---
plt.figure(figsize=(8, 5))
plt.plot(x, t_pdf, label=f"Rozkład t-Studenta (df={df})", color='blue', linewidth=2)
plt.plot(x, norm_pdf, label="Rozkład normalny (Gaussa)", color='orange', linestyle='--', linewidth=2)
plt.plot(x, chi2_pdf, label=f"Rozkład Chi2-Studenta (df={df})", color='green', linewidth=2)
plt.plot(x, f_pdf, label=f"Rozkład F-Studenta (df={df})", color='black', linewidth=2)

plt.title("Porównanie rozkładu t-Studenta i normalnego")
plt.xlabel("Wartość zmiennej")
plt.ylabel("Gęstość prawdopodobieństwa")
plt.legend()
plt.grid(True)
plt.show()
```
