(a) Use the sm.OLS() function to perform a simple linear regression
with mpg as the response and horsepower as the predictor. Use

ii. How strong is the relationship between the predictor and

iii. Is the relationship between the predictor and the response

(b) Plot the response and the predictor in a new set of axes ax. Use

(c) Produce some of diagnostic plots of the least squares regression

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm

from statsmodels.stats.anova import anova_lm
from ISLP.models import summarize, poly
from ISLP.models import ModelSpec as MS
from ISLP import load_data


```

```python
df = pd.read_csv("Auto.csv")
df.head()

df.dtypes
```

```python
#sm.OLS?
X=df["horsepower"]
Y=df["mpg"]


droplist = X[X=="?"].index
X=X.drop(droplist)
Y=Y.drop(droplist)
```

```python
data1 = {"horsepower": X.astype("float"), "mpg": Y}

df1 = pd.DataFrame(data=data1)

df1.head()
```

```python
X1 = sm.add_constant(X)
mod = sm.OLS(Y, X1)
res = mod.fit()

print(res.summary())
```

i. Is there a relationship between the predictor and the response?

Yes we can say that there is relationship based on a coef value which is greater than 0 (that leads us to reject the null hypothesis). Also the F-statistic is high (723) what means that the whole model is valueable. (because of the data variability)


ii. How strong is the relationship between the predictor and the response?

The model has approximately 60% of the data consistency. The R^2 value tells us just that. this value is from range <-1; 1> and when the value is close to 0, that means the model does not tell anything. The higher from 0 the better for the model  (But it can't be too close to 1, because overfitting might occur)

R^2 = 1 - MSEPrdicted/MSErealmodel



iii. Is the relationship between the predictor and the response positive or negative?

Because the coef (slope) value is greater than 0, POSITIVE.


iv. What is the predicted mpg associated with a horsepower of 98? What are the associated 95 % confidence and prediction intervals?


```python
prediction = res.get_prediction(pd.DataFrame({'const':1.0,'horsepower':[98]}))
prediction.predicted_mean
```

```python
prediction.conf_int(alpha=0.05)

# We set the 1-apha = 95% of probability that the real value will be in that range
```

b) Plot the response and the predictor in a new set of axes ax. Use the ax.axline() method or the abline() function defned in the lab to display the least squares regression line.

```python
fig, ax = plt.subplots()

ax.scatter(X, Y, label="Actual val")

intercept, slope = res.params

ax.axline(xy1=(0, intercept), slope=slope, color="red", label="Regression")

ax.set_xlabel("horsepower")
ax.set_ylabel("mpg")

ax.legend()
plt.show()
```

(c) Produce some of diagnostic plots of the least squares regression ft as described in the lab. Comment on any problems you see with the ft.

```python
fitted_val = res.fittedvalues # to wartosc predykcji
resid_vals = res.resid # to wartosc bledu, czyli roznicy wartosci realnej od przewidzianej

fig, ax = plt.subplots()

ax.scatter(fitted_val, resid_vals)
ax.axhline(0,color="red", linestyle="--")
ax.set_xlabel("Fitted values")
ax.set_ylabel("Residuals")
```

Dodatkowy termin:
Heteroskedastyczność — wariancja błędów rośnie wraz z wartością przewidywaną,
co łamie jedno z założeń regresji liniowej (stała wariancja błędów).
