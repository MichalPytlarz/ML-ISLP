```python
mod = smf.glm(formula='default ~ income + balance', data=df, family=sm.families.Binomial()).fit() #create & fit model
print(mod.summary())
```

```python
def boot_fn(dataset):
    mod = smf.glm(formula='default ~ income + balance', data=dataset, family=sm.families.Binomial()).fit()
    coef_i = mod.params.iloc[1]
    coef_b = mod.params.iloc[2]
    return [coef_i, coef_b]

boot_fn(df)
```

```python
# c)
def boot(df, bootSample=None):
    if bootSample == None:
        bootSample = len(df)

    bootSample_i = (np.random.rand(bootSample)*len(df)).astype(int)
    bootSample_i = np.array(bootSample_i)
    bootSample_df = df.iloc[bootSample_i]

    return bootSample_df

coeffs = []
n=100
for i in range(0,n):
    coef_i = boot_fn(boot(df))
    coeffs.append(coef_i)

print(pd.DataFrame(coeffs).mean())
```
