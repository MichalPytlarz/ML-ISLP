```python
import numpy as np
```

```python
arr_x = np.array([278, 230, 208, 224, 446, 365, 360, 292, 245, 245, 296])
arr_y = np.array([277, 211, 205, 235, 424, 345, 341, 287, 219, 200, 254])

d = arr_y - arr_x 
diff_obs = np.mean(d)

diff_sim = []
n_iter = 10000


for i in range(n_iter):
    sign = np.random.choice([-1,1], size=len(d))
    diff_sim.append(np.mean(d * sign))
diff_sim = np.array(diff_sim)

p_val = np.mean(np.abs(diff_sim) >= np.abs(diff_obs))

print("Pval: ", p_val)
```
