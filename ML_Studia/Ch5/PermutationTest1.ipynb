```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

```python
arr1 = np.array([378, 346, 245, 285, 365, 245, 208, 360, 296, 224, 292])
arr2 = np.array([218, 264, 211, 180, 256, 240, 261, 205, 145, 195, 187, 210, 378, 204, 232, 237, 310])

arr = np.append(arr1, arr2)

arr_mean = arr.mean()

n_iter = 10000

real_diff_obs = arr1.mean() - arr2.mean()

arr_s_means = np.array([])
arr_ns_means = np.array([])

diff_simulated = []

for k in range(n_iter):
    sample = np.random.choice(arr, 11, replace=False)
    not_sampled = np.setdiff1d(arr, sample)

    diff_simulated.append(sample.mean() - not_sampled.mean())
diff_simulated = np.array(diff_simulated)


p_val = np.mean(np.abs(diff_simulated) >= np.abs(real_diff_obs))
print("p_val:", p_val)
```
