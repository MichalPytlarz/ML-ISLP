```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 

from ISLP import load_data
from ISLP.models import summarize, ModelSpec, poly
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
df = load_data("Default")
df
```

```python
df["default"] = pd.get_dummies(df["default"], drop_first=True)
```

```python
df["default"] = df["default"].astype(int)
df
```

```python
n = [0.2, 0.4, 0.8]
mean_res_comp = np.zeros(len(n))
for i in range(len(n)):
    df_Train, df_Test = train_test_split(df, test_size=n[i],random_state=0)
    
    X_train = df_Train[["balance", "income"]]
    Y_train = df_Train["default"]
    
    X_test = df_Test[["balance", "income"]]
    Y_test = df_Test["default"]
    
    lreg = LogisticRegression()
    lreg.fit(X_train, Y_train)
    
    y_pred = lreg.predict(X_test)
    mean_res_comp[i] = np.mean((Y_test - y_pred)**2)

    
    # y_pred = (lreg.predict_proba(X_test)[:,1] >= 0.7).astype(int)
    # to jest gdy chcemy sami ustalic od jakiego thresholdu klasyfikujemy
    # predict_proba zwraca nam zwykle prawdopodobienstwo
    # natomiast .predict() zwraca zwykle labelki {0,1}
    print(f"Results for test_size={n[i]}")
    print("Mean:", mean_res_comp[i])
    print("Acc:", accuracy_score(Y_test, y_pred))
    print(confusion_matrix(Y_test, y_pred))

```
